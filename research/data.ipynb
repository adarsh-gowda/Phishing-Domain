```python
# List of all your date columns (from your message)
date_columns = [
    'Order Item Scheduled End Date',
    'Order Planned Start Date',
    'Order Scheduled Start Date',
    'Order Actual Start Date',
    'Order Scheduled End Date',
    'Order Confirmed End Date',
    'Order Actual End Date',
    'Order Scheduled Release Date',
    'Order Actual Release Date',
    'Order Planned End Date',
    'Order Item Planned Delivery Date',
    'Order Item Actual Deliver Date',
    'Planned Order Planned Start Date',
    'Planned Order Opening Date',
    'Order Item Total Commitment Date',
    'Order Item Planned End Date'
]
```

```python
for col in date_columns:
    # Clean and parse
    df[col] = pd.to_datetime(
        df[col].astype(str).str.strip(),
        errors='coerce',
        dayfirst=True,  # still safe, since most are d-m-Y
        infer_datetime_format=True
    )

    # Report and debug
    num_invalid = df[col].isna().sum()
    print(f"🕵️ {col}: {df[col].dtype}, {num_invalid} invalid")

    if num_invalid > 0:
        print("  🔍 Example bad values:")
        print(df[df[col].isna()][col].unique()[:5])
```
